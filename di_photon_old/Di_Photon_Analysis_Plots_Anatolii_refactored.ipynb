```python
def get_ratio_correlated_error(error_numerator, numerator, error_denominator, denominator, correlation_coef=1):
    # calculate errors for ratio of two correlated points. Automatically assumes perfect correlation
    ratio = numerator/denominator
    error_ratio = ratio * np.sqrt((error_numerator/numerator)**2 + (error_denominator/denominator)**2 - 2*correlation_coef*(error_numerator/numerator)*(error_denominator/denominator))
    # error_ratio = np.sqrt((error_numerator/denominator)**2 - numerator*error_denominator**2/denominator**2)
    return error_ratio
```

```python
def get_ratio_uncorrelated_error(a, sigma_a, b, sigma_b):
    f = a / b
    sigma_f = np.sqrt((sigma_a/a)**2 + (sigma_b/b)**2)
    return sigma_f
```

```python
def get_ratio_uncorrelated_error(a, sigma_a, b, sigma_b):
    # f = a / b
    f = b / a
    sigma_f = f * np.sqrt((sigma_a/a)**2 + (sigma_b/b)**2)
    return sigma_f
```

```python
def calculate_kl_divergence(hist1, hist2):

    # Normalize histograms to get probability distributions
    p = hist1 / np.sum(hist1)
    q = hist2 / np.sum(hist2)

    # Avoid division by zero and log of zero
    epsilon = 1e-10
    p = np.clip(p, epsilon, 1.0)
    q = np.clip(q, epsilon, 1.0)

    # Calculate KL divergence
    kl_divergence = np.sum(p * np.log(p / q))

    return kl_divergence
```

```python
def get_stats(positions, photons_in_event, bins):
    photon_counts, bin_edges = np.histogram(positions, weights=photons_in_event, bins=bins, range=(0,1))
    event_counts, _ = np.histogram(positions, bins=bin_edges)

    n = event_counts
    successe_counts, _ = np.histogram(positions, weights=(photons_in_event==2).astype(int), bins=bin_edges)
    p = successe_counts/event_counts
    q = (1-p)
    std = np.sqrt( n * p * q)/event_counts

    return photon_counts/event_counts, std, bin_edges

avg_photons, std, bin_edges = get_stats(positions, photons_in_event, bins)
avg_photons_2, std_2, bin_edges = get_stats(positions_2, photons_in_event_2, bins)
```

```python
avg_photons_list = []
for i in range(100):
    avg_photons, std, bin_edges = get_stats(positions_list[i], photons_in_event[i], bins)
    avg_photons_list.append(avg_photons)
```

```python
avg_photons = np.mean([avg_photons_list[i] for i in range(50)], axis=0)
avg_photons_2 = np.mean([avg_photons_list[i] for i in range(50,100)], axis=0)

std = np.std([avg_photons_list[i] for i in range(3)], axis=0)
std_2 = np.std([avg_photons_list[i] for i in range(3,6)], axis=0)
```

```python
width = np.diff(bin_edges)[0]
bin_centers = bin_edges[:-1] + width/2
```

```python
fig, axes = plt.subplots(2, 1, sharex=True, figsize=(7,9), gridspec_kw={'height_ratios': [3, 1], 'hspace' : 0.1})

axes[0].hist(bin_centers, bins=bin_edges, weights=avg_photons, histtype='step', color='black', lw=1)
axes[0].bar(bin_centers, height=std*2, bottom=avg_photons-std, alpha=0.3, width=width, color='black', linewidth=0)

axes[0].hist(bin_centers, bins=bin_edges, weights=avg_photons_2, histtype='step', color='tab:orange', lw=1)
axes[0].bar(bin_centers, height=std_2*2, bottom=avg_photons_2-std_2, alpha=0.3, width=width, color='tab:orange', linewidth=0)

true_mean = bin_centers+1
axes[0].hist(bin_centers, bins=bin_edges, weights=true_mean, histtype='step', color='blue', lw=1, linestyle='--')


error_R = get_ratio_uncorrelated_error(avg_photons_2, std_2, avg_photons, std)

axes[1].hist(bin_centers, bins=bin_edges, weights=avg_photons/avg_photons_2, histtype='step', color='tab:orange', lw=1)
axes[1].bar(bin_centers, height=error_R*2, bottom=avg_photons/avg_photons_2-error_R, alpha=0.3, width=width, color='tab:orange', linewidth=0)

n_overlap_bins = sum(abs(avg_photons/avg_photons_2 - 1) < error_R)
print(n_overlap_bins)

axes[1].axhline(1, color='black', lw=1.5)
axes[0].set_ylim(0.7, 2.5)
axes[1].set_ylim(0.95, 1.05)

plt.show()
```
