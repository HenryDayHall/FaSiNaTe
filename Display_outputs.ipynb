# Display the plots

Load and display the images produced for a quick verification.

```python
import os
from IPython.display import Image, display
from ddml_bench.config import load_config
import numpy as np
from matplotlib import pyplot as plt
config = load_config()
data_dir = config['paths']['data_dir']
plot_dir = os.path.join(data_dir, 'observables', 'plots')
print(os.listdir(plot_dir))
```

```python

g4_path = os.path.join(data_dir, 'observables', 'geant4.npz')
g4_data = np.load(g4_path)
print(list(g4_data.keys()))
g4_data['energy__single_photon__true']
```

```python

for imageName in sorted(os.listdir(plot_dir)):
    if not imageName.endswith('.png'):
        continue
    path = os.path.join(plot_dir, imageName)
    print(imageName)
    display(Image(filename=path))
```
